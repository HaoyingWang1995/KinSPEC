```python
#!/usr/bin/env python3
"""Download files from unpublished (draft) Zenodo depositions."""

import os
import requests

# ── Fill in your Zenodo personal access token here ──────────────────
TOKEN = "kdsw0bSFhiTzASl6wUxVOqf3YOdijsPKBOTnl3ll4plI7j6AmjkUe6aDvtbk"
# ────────────────────────────────────────────────────────────────────

# (deposition_id, destination_directory)
JOBS = [
    ("21739888", "/home/s1856122/KinSPEC/model"),
    ("20558270", "/home/s1856122/KinSPEC/dataset"),
]

BASE = "https://zenodo.org/api/deposit/depositions"


def download_deposition(dep_id, dest_dir, token):
    os.makedirs(dest_dir, exist_ok=True)
    headers = {"Authorization": f"Bearer {token}"}

    # Get file list for the draft deposition
    r = requests.get(f"{BASE}/{dep_id}", headers=headers, timeout=60)
    r.raise_for_status()
    files = r.json().get("files", [])

    if not files:
        print(f"[{dep_id}] No files found.")
        return

    print(f"[{dep_id}] {len(files)} file(s) → {dest_dir}")
    for f in files:
        name = f["filename"]
        link = f["links"]["download"]
        out_path = os.path.join(dest_dir, name)

        print(f"  ↓ {name}")
        with requests.get(link, headers=headers, stream=True, timeout=300) as resp:
            resp.raise_for_status()
            with open(out_path, "wb") as fh:
                for chunk in resp.iter_content(chunk_size=1 << 20):
                    fh.write(chunk)
        print(f"    saved: {out_path}")


def main():
    if not TOKEN:
        raise SystemExit("Please set your TOKEN at the top of the script.")
    for dep_id, dest in JOBS:
        download_deposition(dep_id, dest, TOKEN)
    print("Done.")


if __name__ == "__main__":
    main()
```
