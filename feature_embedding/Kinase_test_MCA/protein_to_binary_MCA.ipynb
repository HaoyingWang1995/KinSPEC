```python
import pandas as pd
import numpy as np
```

```python
# 23 bit PSM physicochemical features
aa_to_binary = {
    "A" : "10000000001001010010000",
    "C" : "00100000001000101001000",
    "D" : "00000010100100000100100",
    "E" : "00000010100100010000010",
    "F" : "01000000010001001000001",
    "G" : "10000000001000100110000",
    "H" : "00000001100000110000010",
    "I" : "10000000010001001000100",
    "K" : "00000001100010010000010",
    "L" : "10000000010001010000100",
    "M" : "00100000010001010000010",
    "N" : "00001000100000100100100",
    "P" : "00010000001001000101000",
    "Q" : "00001000100000110000010",
    "R" : "00000001100010010000001",
    "S" : "00000100001000100101000",
    "T" : "00000100001000101001000",
    "V" : "10000000010001001001000",
    "W" : "01000000001001001000001",
    "Y" : "01000000001000101000001",
    "-" : "00000000000000000000000"
}
```

```python
file_path = _kp('feature_embedding/Kinase_test_MCA/human_482_typical_kinase_1955bit.csv')
output_file = _kp('feature_embedding/Kinase_test_MCA/generated/human_482_typical_kinase_1955bit.csv')
_KSPath(output_file).parent.mkdir(parents=True, exist_ok=True)  
pro_seq = pd.read_csv(file_path)
```

```python
def sequence_to_binary(sequence):
    if pd.isna(sequence):
        return np.nan
    return ''.join(aa_to_binary[aa] for aa in sequence if aa in aa_to_binary)
```

```python
print(pro_seq['KLIFS'])
```

```python
for seq in pro_seq['KLIFS']:
    if len(seq.strip()) != 85:
        print(seq)
```

```python
pro_seq['Binary_string'] = pro_seq['KLIFS'].apply(sequence_to_binary)
pro_seq.to_csv(output_file,index=False)
print(f"finished")
```
